```python
# Finding the 25th percentile
# sort data first in ascending order using the function sorted()
sort_data=sorted(data)
data_25=len(sort_data)*0.25
data_25
```

```python
# Finding the 75th percentile
data_75=len(sort_data)*0.75
data_75
```

```python
import statistics as st
import numpy as np

array=[]

for _ in range(15):
    no=int(input('Input number:' ))
    array.append(no)
    a=np.array(array)

# Calculate the mean
a_mean = st.mean(a)

#Calculate the median
a_median=st.median(a)

#Calculate Mode
a_mode=st.mode(a)

# Maximum and Minimum values
a_min=min(a)
a_max=max(a)

#Calculate Range
a_range=max(a)-min(a)

# Finding the interquartile range
# Calculate the 25th percentile
a_sorted=sorted(a)
a_25=len(a_sorted)*0.25

# Calculate the 75th percentile
a_75=len(a_sorted)*0.75

# IQR
IQR=a_75-a_25

# 22, 21, 23, 28, 23, 34, 30, 24, 49, 27, 23, 22, 32, 28, 26
print('Array:', a)
print('Mean:',a_mean)
print('Median:', a_median)
print('Mode:', a_mode)
print('The range is between', a_min, 'and', a_max, 'which is', a_range)
print('The 25th percentile:', a_25)
print('The 75th percentile:', a_75)
print('Your IQR:', IQR)
```
